```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.read_csv('titanic_with_labels.csv', index_col=0) # sep = '', delimiter = '', nrows = 10, header = None
df2 = pd.read_csv('cinema_sessions.csv', index_col=0) # sep = '', delimiter = '', nrows = 10, header = None
df1 = pd.merge(df1, df2, on='check_number')
df1['session_start'] = pd.to_datetime(df1['session_start'], format= 'mixed')
```

```python
groups_bin = [0, 18, 50, np.inf]
labels = ['kid', 'adult', 'elderly']

# Разбиваем возраст на категории
df1['age_group'] = pd.cut(df1['age'], bins=groups_bin, labels=labels, include_lowest=True)

# Создаём бинарные колонки для каждой категории
df1['age_child']   = np.where(df1['age_group'] == 'kid', 1, 0)
df1['age_adult']   = np.where(df1['age_group'] == 'adult', 1, 0)
df1['age_elderly'] = np.where(df1['age_group'] == 'elderly', 1, 0)

# Удаляем исходный столбец возраста и категорию
df1 = df1.drop(['age', 'age_group'], axis=1)

df1
```

```python
df1['drink'] = df1['drink'].map({'Beerbeer': 1, 'Water': 0, 'Bugbeer' : 1, 'Cola' : 0, 'Strong beer' : 1, 'Fanta' : 0, 'Наше пиво' : 1})
```

```python
# df1['session_start'].max()
bins = [6, 12, 20, 24]
labels = ['morning', 'day', 'evening']

df1['session_start_group'] = pd.cut(df1['session_start'].dt.hour, bins= bins, labels= labels, include_lowest= True)
time_dummies = pd.get_dummies(df1['session_start_group'], prefix='session')
df1 = pd.concat([df1, time_dummies], axis=1)

df1 = df1.drop(['session_start_group'], axis=1)

for col in ['session_morning', 'session_day', 'session_evening']:
    df1[col] = df1[col].astype(int)
```
