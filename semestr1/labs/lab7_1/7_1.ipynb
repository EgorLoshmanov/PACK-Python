```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('titanic_with_labels.csv', index_col=0) # sep = '', delimiter = '', nrows = 10, header = None
```

```python
df = df[~df['sex'].isin(['Не указан', "не указан", '-'])].copy()  
df.loc[:, 'sex'] = df['sex'].map({'M': 0, 'Мужчина': 0,'F': 1})
```

```python
df['row_number'] = df['row_number'].fillna(df['row_number'].max())
```

```python
q1 = df['liters_drunk'].quantile(0.25)
q3 = df['liters_drunk'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
```

```python
df['liters_drunk'] = df['liters_drunk'].astype(float)

valid_mask = (df['liters_drunk'] >= 0) & (df['liters_drunk'] >= lower_bound) & (df['liters_drunk'] <= upper_bound)
mean_valid = df.loc[valid_mask, 'liters_drunk'].mean()

df.loc[~valid_mask, 'liters_drunk'] = mean_valid
```
